## Task 1: Text Preprocessing Pipeline
Remove this cell if you're not completing this task.

```python
import re
import matplotlib.pyplot as plt
from collections import Counter
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download('stopwords', quiet=True)

#first_task_text_preprocessing pipeline

def preprocess_text(text):
    """
    This applies a preprocessing pipeline:
    1. Lowercase
    2. Remove punctuation/special characters
    3. Tokenization
    4. Remove stopwords
    5. Stemming
    """
    # converts_text_to_lowercase_text
    text = text.lower()
    
    #regex_removing_punctuations_and_special_characters
    text = re.sub(r'[^a-z\s]', '', text)
    
    #tokenization
    tokens = text.split()
    
    #removes_stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    
    #stemming
    stemmer = PorterStemmer()
    processed_tokens = [stemmer.stem(word) for word in tokens]
    
    return processed_tokens

#testing_and_visualization
raw_text = """
Natural Language Processing (NLP) is largely a very interesting field! In all AI domains NLP stands out (don't be jealous other domains)! It could save languages!
NLP helps linguistics and computer science (a very intersting intersection) to help machines understand human languages. And how powerful it is when machines understand human language!
"""

# applies_pipeline
clean_tokens = preprocess_text(raw_text)

# Show Before/After
print(f"ORIGINAL:\n{raw_text.strip()}\n")
print(f"PROCESSED TOKENS:\n{clean_tokens}\n")

# Count frequencies
word_counts = Counter(clean_tokens)

# Visualize Top 5 Words
common_words = word_counts.most_common(5)
words, counts = zip(*common_words)

plt.figure(figsize=(8, 4))
plt.bar(words, counts, color='teal')
plt.title("Task 1: Top Word Frequencies")
plt.xlabel("Stems")
plt.ylabel("Count")
plt.show()
```

## Task 2: Simple Text Classifier
Remove this cell if you're not completing this task.

```python
import re
import string
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download('stopwords', quiet=True)
print("The Libraries are imported and the NLTK data is sure ready.")

#le_text_classfier (; p.s the french is just for an uninvited daramtic effect! ;)

class SimpleSentimentClassifier:
    def __init__(self):
        self.positive_words = {'good', 'great', 'love', 'excellent', 'amazing', 'best', 'happy', 'like'}
        self.negative_words = {'bad', 'terrible', 'hate', 'poor', 'awful', 'worst', 'sad', 'disappointed'}

    def predict(self, text):
        text = text.lower()
        text = re.sub(r'[^a-z\s]', '', text)
        tokens = text.split()
        
        #scoring
        score = 0
        for word in tokens:
            if word in self.positive_words:
                score += 1
            elif word in self.negative_words:
                score -= 1
        
        #making_the_ve_decisions
        if score > 0:
            return "Positive"
        elif score < 0:
            return "Negative"
        else:
            return "Neutral"
        
#testinggg
test_data = [
    ("I love this watch, it is amazing!", "Positive"),
    ("Dracula is the worst movie I have ever seen.", "Negative"),
    ("The Abuja weather is okay, not great but not bad.", "Neutral"),
    ("I am very happy with the excellent services in Nile University", "Positive"),
    ("Terrible food, I hate poor cooking.", "Negative"),
    ("The visit to the zoo where I carried a lion was poor and the service was awful. The cub scraped my hand with its paws", "Negative"),
    ("I like the Making it Big book by Femi Otedola, it is good.", "Positive"),
    ("It is just a regular book.", "Neutral"),
    ("Absolutely disappointed, very sad result.", "Negative"),
    ("Best day ever!", "Positive")
]

classifier = SimpleSentimentClassifier()
correct_count = 0

print(f"{'PREDICTION':<12} | {'ACTUAL':<12} | {'TEXT'}")
print("-" * 60)

for text, actual in test_data:
    prediction = classifier.predict(text)
    if prediction == actual:
        correct_count += 1
    print(f"{prediction:<12} | {actual:<12} | {text}")

accuracy = (correct_count / len(test_data)) * 100
print("-" * 60)
print(f"Model Accuracy: {accuracy}%")
```
