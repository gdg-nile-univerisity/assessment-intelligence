## Task 2: Simple Image Filter
Remove this cell if you're not completing this task.

```python
# Your code for Task 2 here
imagearray = image.astype(np.float32)

# Add brightness
brightness = imagearray + 5
brightnesseffect = np.clip(brightness, 0, 255)
brightnesseffect = brightnesseffect.astype(np.uint8)

# Adjust contrast
mean = np.mean(brightnesseffect, axis=(0,1), keepdims=True)
contrast = (brightnesseffect - mean) * 0.5 + mean
contrasteffect = np.clip(contrast, 0, 255)
contrasteffect = contrasteffect.astype(np.uint8)

# Blur
blureffect = cv.blur(contrasteffect, (8,8))
final = cv.cvtColor(blureffect, cv.COLOR_BGR2RGB)

plt.imshow(final)
plt.axis("off")
plt.show()
```
